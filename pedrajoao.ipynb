```python
import random
import socket

laptop = socket.gethostname()

op = ["pedra", "papel", "tesoura"]
v = 0
d = 0

print(f"[{laptop}] Pedra, Papel e Tesoura - Melhor de 3!")

while v < 2 and d < 2:
    voce = input("pedra/papel/tesoura: ").strip().lower()

    if voce not in op:
        print("Inválido! Digite pedra, papel ou tesoura.")
        continue

    pc = random.choice(op)

    print(f"Você: {voce} | PC: {pc}")

    if voce == pc:
        print("Empate!")

    elif (voce, pc) in [
        ("pedra", "tesoura"),
        ("papel", "pedra"),
        ("tesoura", "papel")
    ]:
        v += 1
        print("Ponto seu!")

    else:
        d += 1
        print("Ponto do PC!")

    print(f"Placar: {v} x {d}")
    print("-" * 30)

if v > d:
    print("🏆 VOCÊ VENCEU!")
else:
    print("🤖 PC VENCEU!")
```
