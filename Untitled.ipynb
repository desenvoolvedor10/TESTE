```python
# ============================================================
# QUIZ RELÂMPAGO
# ============================================================

# Importa a biblioteca socket
import socket


# ------------------------------------------------------------
# CONFIGURAÇÃO INICIAL
# ------------------------------------------------------------

# Descobre o nome do computador
laptop = socket.gethostname()

# Perguntas e respostas
perg = [
    ("Quanto é 7 x 8?", "56"),
    ("Capital do Brasil (sem acento)?", "brasilia"),
    ("Função que mostra texto na tela?", "print"),
    ("Quanto é 2 ** 3?", "8")
]

# Inicializa a pontuação
pontos = 0


# ------------------------------------------------------------
# INÍCIO DO QUIZ
# ------------------------------------------------------------

print("=" * 50)
print(f"[{laptop}] QUIZ RELÂMPAGO!")
print("=" * 50)

# Percorre todas as perguntas
for i, (p, resp) in enumerate(perg, 1):

    resposta = input(f"{i}) {p} ").strip().lower()

    # Verifica se a resposta está correta
    if resposta == resp:
        print("✅ Certo!")
        pontos += 1
    else:
        print("❌ Errado!")
        print(f"A resposta correta era: {resp}")

    print()


# ------------------------------------------------------------
# RESULTADO FINAL
# ------------------------------------------------------------

print("=" * 50)
print("RESULTADO FINAL")
print("=" * 50)

print(f"Você fez {pontos} de {len(perg)} pontos!")

if pontos == len(perg):
    print("🏆 Parabéns! Você acertou tudo!")

elif pontos >= 2:
    print("👏 Muito bem!")

else:
    print("📚 Continue praticando!")

print("=" * 50)

```
