```python
/* ---------- LOGIN (localStorage) ---------- */
var modo="entrar";
function users(){ try{return JSON.parse(localStorage.getItem("aegis_users")||"{}");}catch(e){return {};} }
function salvaUsers(u){ localStorage.setItem("aegis_users",JSON.stringify(u)); }
function troca(m){
  modo=m;
  document.getElementById("tabEntrar").classList.toggle("on",m==="entrar");
  document.getElementById("tabCadastrar").classList.toggle("on",m==="cadastrar");
  document.getElementById("btnAcao").textContent = m==="entrar"?"Entrar":"Cadastrar";
  setMsg("","");
}
function setMsg(t,cls){ var e=document.getElementById("msg"); e.textContent=t; e.className="msg "+(cls||""); }
function acao(){
  var email=document.getElementById("email").value.trim().toLowerCase();
  var senha=document.getElementById("senha").value;
  if(!email || !senha){ setMsg("Preencha e-mail e senha.","err"); return; }
  var u=users();
  if(modo==="cadastrar"){
    if(u[email]){ setMsg("Este e-mail ja tem cadastro. Clique em Entrar.","err"); return; }
    u[email]=senha; salvaUsers(u);
    setMsg("Cadastro criado! Agora clique em Entrar.","ok"); troca("entrar");
  }else{
    if(!u[email]){ setMsg("E-mail nao cadastrado. Clique em Cadastrar.","err"); return; }
    if(u[email]!==senha){ setMsg("Senha incorreta.","err"); return; }
    entrar(email);
  }
}
function entrar(email){
  sessionStorage.setItem("aegis_user",email);
  document.getElementById("login").style.display="none";
  document.getElementById("app").hidden=false;
  document.getElementById("who").textContent="Logado: "+email;
  vNovo(); fNovo(); cInit();
}
function sair(){
  sessionStorage.removeItem("aegis_user");
  document.getElementById("app").hidden=true;
  document.getElementById("login").style.display="flex";
  document.getElementById("senha").value=""; setMsg("","");
}
function abre(p,el){
  document.querySelectorAll(".page").forEach(function(s){s.classList.remove("on");});
  document.getElementById(p).classList.add("on");
  document.querySelectorAll("nav a").forEach(function(a){a.classList.remove("on");});
  el.classList.add("on");
}
```

```python
/* ---------- JOGO DA VELHA ---------- */
var vTab,vVez,vFim;
function vNovo(){
  vTab=["","","","","","","","",""]; vVez="X"; vFim=false;
  document.getElementById("vStatus").textContent="Vez do jogador X";
  var g=document.getElementById("vGrid"); g.innerHTML="";
  for(var i=0;i<9;i++){ (function(i){
    var b=document.createElement("button"); b.className="cell"; b.onclick=function(){vJoga(i);};
    g.appendChild(b);
  })(i); }
}
function vJoga(i){
  if(vFim || vTab[i]) return;
  vTab[i]=vVez; document.getElementById("vGrid").children[i].textContent=vVez;
  var linhas=[[0,1,2],[3,4,5],[6,7,8],[0,3,6],[1,4,7],[2,5,8],[0,4,8],[2,4,6]];
  for(var k=0;k<linhas.length;k++){ var a=linhas[k];
    if(vTab[a[0]] && vTab[a[0]]===vTab[a[1]] && vTab[a[1]]===vTab[a[2]]){
      document.getElementById("vStatus").textContent="Jogador "+vVez+" venceu!"; vFim=true; return; } }
  if(vTab.indexOf("")===-1){ document.getElementById("vStatus").textContent="Deu velha! (empate)"; vFim=true; return; }
  vVez = vVez==="X"?"O":"X";
  document.getElementById("vStatus").textContent="Vez do jogador "+vVez;
}
```

```python
import http.server, socketserver, threading, webbrowser, os, functool

/* ---------- JOGO DA FORCA ---------- */
var fPal,fAcertos,fErros;
var fLista=["python","jupyter","codigo","aegishub","notebook","programar","internet","teclado","projeto","futuro"];
function fNovo(){
  fPal=fLista[Math.floor(Math.random()*fLista.length)]; fAcertos=[]; fErros=0;
  document.getElementById("fStatus").textContent="Boa sorte!";
  fDesenha();
  var kb=document.getElementById("fKeys"); kb.innerHTML="";
  "abcdefghijklmnopqrstuvwxyz".split("").forEach(function(l){
    var b=document.createElement("button"); b.textContent=l; b.onclick=function(){fLetra(l,b);};
    kb.appendChild(b);
  });
}
function fDesenha(){
  document.getElementById("fLives").textContent="Erros: "+fErros+" / 6";
  var out=fPal.split("").map(function(c){return fAcertos.indexOf(c)>=0?c:"_";}).join(" ");
  document.getElementById("fWord").textContent=out;
}
function fLetra(l,btn){
  btn.disabled=true;
  if(fPal.indexOf(l)>=0){ fAcertos.push(l); } else { fErros++; }
  fDesenha();
  var ganhou=fPal.split("").every(function(c){return fAcertos.indexOf(c)>=0;});
  if(ganhou){ document.getElementById("fStatus").textContent="PARABENS! Voce acertou: "+fPal; fTrava(); }
  else if(fErros>=6){ document.getElementById("fStatus").textContent="Fim! A palavra era: "+fPal;
    document.getElementById("fWord").textContent=fPal.split("").join(" "); fTrava(); }
}
function fTrava(){ document.querySelectorAll("#fKeys button").forEach(function(b){b.disabled=true;}); }

/* ---------- CALCULADORA ---------- */
var cAtual,cAnterior,cOp;
function cInit(){
  cLimpa();
  var pad=document.getElementById("cPad"); pad.innerHTML="";
  var teclas=[["C","clr"],["/","op"],["*","op"],["<","op"],
              ["7",""],["8",""],["9",""],["-","op"],
              ["4",""],["5",""],["6",""],["+","op"],
              ["1",""],["2",""],["3",""],["=","eq"],
              ["0",""],[".",""]];
  teclas.forEach(function(t){
    var b=document.createElement("button"); b.textContent=t[0]; if(t[1])b.className=t[1];
    b.onclick=function(){cTecla(t[0]);}; pad.appendChild(b);
  });
}
function cMostra(v){ document.getElementById("cDisp").textContent=String(v).slice(0,12); }
function cLimpa(){ cAtual="0"; cAnterior=null; cOp=null; cMostra("0"); }
function cTecla(k){
  if(k==="C"){ cLimpa(); return; }
  if(k==="<"){ cAtual=cAtual.length>1?cAtual.slice(0,-1):"0"; cMostra(cAtual); return; }
  if(k>="0"&&k<="9"){ cAtual = cAtual==="0"?k:cAtual+k; cMostra(cAtual); return; }
  if(k==="."){ if(cAtual.indexOf(".")<0)cAtual+="."; cMostra(cAtual); return; }
  if(k==="+"||k==="-"||k==="*"||k==="/"){
    if(cOp&&cAnterior!==null){ cIgual(); } else { cAnterior=parseFloat(cAtual); }
    cOp=k; cAtual="0"; return;
  }
  if(k==="="){ cIgual(); }
}
function cIgual(){
  if(cOp===null||cAnterior===null) return;
  var b=parseFloat(cAtual),r=0;
  if(cOp==="+")r=cAnterior+b; else if(cOp==="-")r=cAnterior-b;
  else if(cOp==="*")r=cAnterior*b; else if(cOp==="/")r=(b===0?"Erro":cAnterior/b);
  cMostra(r); cAtual=String(r==="Erro"?0:Math.round(r*1e8)/1e8); cAnterior=null; cOp=null;
}
</script>
</body>
</html>
"""

pasta = os.path.abspath("meu_app_web")
os.makedirs(pasta, exist_ok=True)
with open(os.path.join(pasta, "index.html"), "w", encoding="utf-8") as f:
    f.write(HTML)

Handler = functools.partial(http.server.SimpleHTTPRequestHandler, directory=pasta)
httpd = None
for port in range(8501, 8511):
    try:
        httpd = socketserver.TCPServer(("127.0.0.1", port), Handler); break
    except OSError:
        httpd = None
if httpd is None:
    httpd = socketserver.TCPServer(("127.0.0.1", 0), Handler); port = httpd.server_address[1]

threading.Thread(target=httpd.serve_forever, daemon=True).start()
url = "http://127.0.0.1:%d/index.html" % port
print("=" * 46)
print(" SEU APP WEB ESTA NO AR!")
print(" Abra no navegador:", url)
print("=" * 46)
try:
    webbrowser.open(url)
except Exception:
    pass
```
