```python
import random, socket
laptop = socket.gethostname()
palavras = ["cachorro", "gato", "boi", "vaca", "tatu"]
palavra = random.choice(palavras)
acertadas = set()
erros = 0
max_erros = 6
print(f"[{laptop}] Forca - adivinhe a palavra! Voce pode errar {max_erros} vezes.")
while erros < max_erros:
    mostra = " ".join([c if c in acertadas else "_" for c in palavra])
    print("Palavra:", mostra)
    if all(c in acertadas for c in palavra):
        print("PARABENS! Voce adivinhou:", palavra)
        break
    letra = input("Diga uma letra: ").strip().lower()
    if not letra:
        continue
    letra = letra[0]
    if letra in palavra:
        acertadas.add(letra)
        print("Boa, tem essa letra!")
    else:
        erros += 1
        print(f"Nao tem. Erros: {erros}/{max_erros}")
else:
    print("Fim de jogo! A palavra era:", palavra)
```
