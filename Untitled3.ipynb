```python
import random
import socket

# ============================================================
# 🏆 CAMPEONATO DE PROGRAMACAO - PYTHON
# ============================================================

laptop = socket.gethostname()

print("=" * 65)
print("🏆 CAMPEONATO DE PROGRAMACAO - PYTHON")
print(f"💻 Computador: {laptop}")
print("=" * 65)


# ============================================================
# FUNCAO - JOGO ADIVINHE O NUMERO
# ============================================================

def jogo_numero(nome):

    print("\n" + "=" * 65)
    print(f"🎯 {nome.upper()} - ADIVINHE O NUMERO")
    print("=" * 65)

    segredo = random.randint(1, 20)

    print("O computador escolheu um numero de 1 a 20.")
    print("Voce tem 5 tentativas!")

    for tentativa in range(1, 6):

        while True:
            try:
                palpite = int(
                    input(f"\nTentativa {tentativa}/5 - seu numero: ")
                )

                if 1 <= palpite <= 20:
                    break

                print("⚠️ Digite um numero entre 1 e 20.")

            except ValueError:
                print("⚠️ Digite somente numeros.")

        if palpite == segredo:

            pontos = 60 - ((tentativa - 1) * 10)

            print("\n🎉 ACERTOU!")
            print(f"⭐ +{pontos} pontos")

            return True, pontos

        elif palpite < segredo:
            print("⬆️ O numero e MAIOR.")

        else:
            print("⬇️ O numero e MENOR.")

    print("\n❌ Acabaram as tentativas!")
    print(f"O numero correto era {segredo}.")

    return False, 0


# ============================================================
# FUNCAO - JOGO DA FORCA
# ============================================================

def jogo_forca(nome):

    palavras = [
        "python",
        "jupyter",
        "codigo",
        "aegishub",
        "notebook",
        "algoritmo",
        "programacao"
    ]

    palavra = random.choice(palavras)

    acertadas = set()
    tentadas = set()

    erros = 0
    max_erros = 6

    print("\n" + "=" * 65)
    print(f"🔓 {nome.upper()} - JOGO DA FORCA")
    print("=" * 65)

    print(f"Voce pode errar {max_erros} vezes.")

    while erros < max_erros:

        mostra = " ".join(
            letra if letra in acertadas else "_"
            for letra in palavra
        )

        print("\nPalavra:", mostra)
        print(f"❌ Erros: {erros}/{max_erros}")

        if tentadas:
            print("🔤 Letras tentadas:", " ".join(sorted(tentadas)))

        # Verifica se ganhou
        if all(letra in acertadas for letra in palavra):

            pontos = 60 - (erros * 10)

            print("\n🏆 PARABENS!")
            print("Voce descobriu:", palavra.upper())
            print(f"⭐ +{pontos} pontos")

            return True, pontos

        letra = input("\nDigite uma letra: ").strip().lower()

        if not letra:
            print("⚠️ Digite uma letra.")
            continue

        letra = letra[0]

        if not letra.isalpha():
            print("⚠️ Digite uma letra valida.")
            continue

        if letra in tentadas:
            print("⚠️ Voce ja tentou essa letra.")
            continue

        tentadas.add(letra)

        if letra in palavra:

            acertadas.add(letra)
            print("✅ Boa! Essa letra existe.")

        else:

            erros += 1
            print("❌ Essa letra nao existe.")

    print("\n💀 FIM DE JOGO!")
    print("A palavra era:", palavra.upper())

    return False, 0


# ============================================================
# CADASTRO DOS COMPETIDORES
# ============================================================

print("\n📝 CADASTRO DOS COMPETIDORES")

while True:

    try:
        quantidade = int(
            input("\nQuantos competidores participarao? ")
        )

        if quantidade >= 2:
            break

        print("⚠️ Precisamos de pelo menos 2 competidores.")

    except ValueError:
        print("⚠️ Digite um numero valido.")


competidores = []

for numero in range(1, quantidade + 1):

    while True:

        nome = input(
            f"Nome do competidor {numero}: "
        ).strip()

        if nome:
            break

        print("⚠️ Digite um nome.")

    competidores.append({
        "nome": nome,
        "pontos": 0
    })


# ============================================================
# MOSTRA OS COMPETIDORES
# ============================================================

print("\n" + "=" * 65)
print("👥 COMPETIDORES")
print("=" * 65)

for numero, jogador in enumerate(competidores, start=1):
    print(f"{numero}. {jogador['nome']}")


input("\nPressione ENTER para iniciar o campeonato...")


# ============================================================
# 🔥 FASE CLASSIFICATORIA
# ============================================================

print("\n" + "=" * 65)
print("🔥 FASE CLASSIFICATORIA")
print("=" * 65)

for jogador in competidores:

    nome = jogador["nome"]

    print("\n\n" + "🏁" * 25)
    print(f"\n🎮 VEZ DE {nome.upper()}")
    print("\n" + "🏁" * 25)

    input("\nPressione ENTER quando estiver pronto...")


    # --------------------------------------------------------
    # JOGO 1
    # --------------------------------------------------------

    venceu_numero, pontos_numero = jogo_numero(nome)

    jogador["pontos"] += pontos_numero


    # --------------------------------------------------------
    # JOGO 2 - SOMENTE SE ACERTAR O NUMERO
    # --------------------------------------------------------

    if venceu_numero:

        print("\n🔓 PARABENS!")
        print("VOCE DESBLOQUEOU O JOGO DA FORCA!")

        input("\nPressione ENTER para jogar a FORCA...")

        venceu_forca, pontos_forca = jogo_forca(nome)

        jogador["pontos"] += pontos_forca

    else:

        print("\n🔒 JOGO DA FORCA BLOQUEADO!")
        print("Era necessario vencer o Adivinhe o Numero.")


    # --------------------------------------------------------
    # RESULTADO
    # --------------------------------------------------------

    print("\n" + "-" * 65)
    print(f"🏁 RESULTADO DE {nome.upper()}")
    print(f"⭐ TOTAL: {jogador['pontos']} PONTOS")
    print("-" * 65)

    input("\nPressione ENTER para o proximo competidor...")


# ============================================================
# 🏆 RANKING DA CLASSIFICATORIA
# ============================================================

ranking = sorted(
    competidores,
    key=lambda jogador: jogador["pontos"],
    reverse=True
)

print("\n\n" + "=" * 65)
print("🏆 RANKING DA FASE CLASSIFICATORIA")
print("=" * 65)

for posicao, jogador in enumerate(ranking, start=1):

    if posicao == 1:
        medalha = "🥇"

    elif posicao == 2:
        medalha = "🥈"

    elif posicao == 3:
        medalha = "🥉"

    else:
        medalha = "  "

    print(
        f"{medalha} {posicao}º - "
        f"{jogador['nome']} "
        f"| ⭐ {jogador['pontos']} pontos"
    )


# ============================================================
# 🔥 SELECIONA OS DOIS FINALISTAS
# ============================================================

finalista1 = ranking[0]
finalista2 = ranking[1]

print("\n" + "=" * 65)
print("🔥🔥🔥 CLASSIFICADOS PARA A GRANDE FINAL 🔥🔥🔥")
print("=" * 65)

print(
    f"\n🥇 {finalista1['nome'].upper()} "
    f"- {finalista1['pontos']} pontos"
)

print(
    f"🥈 {finalista2['nome'].upper()} "
    f"- {finalista2['pontos']} pontos"
)

print("\n⚔️ GRANDE FINAL")

print(f"""
        🔥 {finalista1['nome'].upper()}
                 VS
        🔥 {finalista2['nome'].upper()}
""")

input("Pressione ENTER para iniciar a GRANDE FINAL...")


# ============================================================
# 🏆 GRANDE FINAL
# ============================================================

print("\n\n" + "=" * 65)
print("🏆🏆🏆 GRANDE FINAL 🏆🏆🏆")
print("=" * 65)

print("\nNa FINAL os pontos da classificatoria sao zerados.")
print("Cada finalista joga novamente:")
print("🎯 Adivinhe o Numero")
print("🔓 Se vencer, joga a Forca")
print("\nQuem fizer mais pontos sera o CAMPEAO!")


finalistas = [
    {
        "nome": finalista1["nome"],
        "pontos_final": 0
    },
    {
        "nome": finalista2["nome"],
        "pontos_final": 0
    }
]


# ============================================================
# JOGOS DA FINAL
# ============================================================

for jogador in finalistas:

    nome = jogador["nome"]

    print("\n\n" + "🔥" * 25)

    print(f"\n🏆 FINALISTA: {nome.upper()}")

    print("\n" + "🔥" * 25)

    input("\nPressione ENTER quando estiver pronto...")


    # --------------------------------------------------------
    # ADIVINHE O NUMERO - FINAL
    # --------------------------------------------------------

    venceu_numero, pontos_numero = jogo_numero(nome)

    jogador["pontos_final"] += pontos_numero


    # --------------------------------------------------------
    # FORCA - FINAL
    # --------------------------------------------------------

    if venceu_numero:

        print("\n🔓 FORCA DA FINAL DESBLOQUEADA!")

        input("\nPressione ENTER para continuar...")

        venceu_forca, pontos_forca = jogo_forca(nome)

        jogador["pontos_final"] += pontos_forca

    else:

        print("\n🔒 FORCA DA FINAL BLOQUEADA!")


    print("\n" + "-" * 65)

    print(
        f"⭐ {nome.upper()} TERMINOU A FINAL "
        f"COM {jogador['pontos_final']} PONTOS"
    )

    print("-" * 65)

    input("\nPressione ENTER para continuar...")


# ============================================================
# RESULTADO DA FINAL
# ============================================================

f1 = finalistas[0]
f2 = finalistas[1]

print("\n\n" + "=" * 65)
print("📊 RESULTADO DA GRANDE FINAL")
print("=" * 65)

print(
    f"\n🔥 {f1['nome']}: "
    f"{f1['pontos_final']} pontos"
)

print(
    f"🔥 {f2['nome']}: "
    f"{f2['pontos_final']} pontos"
)


# ============================================================
# DEFINE O CAMPEAO
# ============================================================

if f1["pontos_final"] > f2["pontos_final"]:

    campeao = f1
    vice = f2

elif f2["pontos_final"] > f1["pontos_final"]:

    campeao = f2
    vice = f1

else:

    # ========================================================
    # DESEMPATE
    # ========================================================

    print("\n⚠️ EMPATE NA GRANDE FINAL!")
    print("🔥 TEREMOS MORTE SUBITA!")
    print("Quem acertar o numero em menos tentativas vence.")

    while True:

        resultados_desempate = []

        for jogador in finalistas:

            nome = jogador["nome"]

            print("\n" + "=" * 65)
            print(f"⚡ MORTE SUBITA - {nome.upper()}")
            print("=" * 65)

            segredo = random.randint(1, 20)

            tentativa_acerto = 999

            for tentativa in range(1, 6):

                while True:

                    try:

                        palpite = int(
                            input(
                                f"Tentativa {tentativa}/5 "
                                f"- numero: "
                            )
                        )

                        if 1 <= palpite <= 20:
                            break

                        print("Digite um numero entre 1 e 20.")

                    except ValueError:

                        print("Digite somente numeros.")

                if palpite == segredo:

                    print("🎯 ACERTOU!")

                    tentativa_acerto = tentativa

                    break

                elif palpite < segredo:

                    print("⬆️ MAIOR")

                else:

                    print("⬇️ MENOR")

            if tentativa_acerto == 999:

                print(f"O numero era {segredo}.")

            resultados_desempate.append({
                "nome": nome,
                "tentativas": tentativa_acerto
            })


        resultado1 = resultados_desempate[0]
        resultado2 = resultados_desempate[1]


        if resultado1["tentativas"] < resultado2["tentativas"]:

            campeao = finalistas[0]
            vice = finalistas[1]

            break

        elif resultado2["tentativas"] < resultado1["tentativas"]:

            campeao = finalistas[1]
            vice = finalistas[0]

            break

        else:

            print("\n😱 EMPATARAM NOVAMENTE!")
            print("🔥 NOVA MORTE SUBITA!")


# ============================================================
# 👑 CAMPEAO
# ============================================================

print("\n\n")
print("🏆" * 25)

print("\n            TEMOS UM CAMPEAO!\n")

print(f"        👑 {campeao['nome'].upper()} 👑")

print("\n      🏆 CAMPEAO DO DESAFIO PYTHON 🏆")

print(f"\n🥈 Vice-campeao: {vice['nome']}")

print("\n" + "🏆" * 25)

print("\n🎉 PARABENS A TODOS OS COMPETIDORES!")
print("💻 Continuem programando!")
```
