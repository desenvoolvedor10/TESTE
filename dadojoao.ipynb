```python
import random
import socket

laptop = socket.gethostname()
voce = 0
pc = 0

print(f"[{laptop}] Duelo de Dados - 3 rodadas!")

for r in range(1, 4):
    input(f"Rodada {r} - ENTER para rolar...")

    a = random.randint(1, 6)
    b = random.randint(1, 6)

    print(f"Você: {a} | PC: {b}")

    if a > b:
        voce += 1
        print("Ponto seu!")

    elif b > a:
        pc += 1
        print("Ponto do PC!")

    else:
        print("Empate!")

    print(f"Placar: {voce} x {pc}")
    print("-" * 30)

if voce > pc:
    print("🏆 VITÓRIA!")

elif pc > voce:
    print("🤖 DERROTA!")

else:
    print("🤝 EMPATE!")
```
