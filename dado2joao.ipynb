```python
# ============================================================
# JOGO: DUELO DE DADOS
# Jogador x Computador
# ============================================================

# Importa a biblioteca random.
# Ela será usada para sortear números aleatórios de 1 até 6.
import random

# Importa a biblioteca socket.
# Ela permite descobrir o nome do computador onde o jogo está rodando.
import socket


# ------------------------------------------------------------
# CONFIGURAÇÃO INICIAL
# ------------------------------------------------------------

# Descobre o nome do laptop/computador.
laptop = socket.gethostname()

# Cria a variável que guarda os pontos do jogador.
# O jogador começa com zero pontos.
voce = 0

# Cria a variável que guarda os pontos do computador.
# O computador também começa com zero pontos.
pc = 0


# ------------------------------------------------------------
# INÍCIO DO JOGO
# ------------------------------------------------------------

# Mostra o nome do computador e o título do jogo.
print("=" * 60)
print("JOGO: DUELO DE DADOS")
print("Jogador x Computador")
print(f"Computador: {laptop}")
print("=" * 60)

print("O jogo terá 3 rodadas!")
print()


# ------------------------------------------------------------
# RODADAS
# ------------------------------------------------------------

# O comando range(1, 4) cria as rodadas 1, 2 e 3.
for rodada in range(1, 4):

    print(f"--- RODADA {rodada} ---")

    # Espera o jogador apertar ENTER.
    input("Pressione ENTER para rolar os dados...")

    # Sorteia um número de 1 até 6 para o jogador.
    dado_voce = random.randint(1, 6)

    # Sorteia um número de 1 até 6 para o computador.
    dado_pc = random.randint(1, 6)

    # Mostra os resultados dos dados.
    print(f"Você tirou: {dado_voce}")
    print(f"PC tirou: {dado_pc}")

    # Verifica quem tirou o maior número.
    if dado_voce > dado_pc:
        voce += 1
        print("🎉 Ponto para você!")

    elif dado_pc > dado_voce:
        pc += 1
        print("🤖 Ponto para o PC!")

    else:
        print("🤝 Empate! Ninguém ganhou ponto.")

    # Mostra o placar atual.
    print(f"Placar: Você {voce} x {pc} PC")
    print()


# ------------------------------------------------------------
# RESULTADO FINAL
# ------------------------------------------------------------

print("=" * 60)
print("RESULTADO FINAL")
print("=" * 60)

print(f"Você: {voce} ponto(s)")
print(f"PC: {pc} ponto(s)")
print()

# Verifica quem venceu o jogo.
if voce > pc:
    print("🏆 VITÓRIA! Você venceu o duelo!")

elif pc > voce:
    print("🤖 DERROTA! O PC venceu o duelo!")

else:
    print("🤝 EMPATE! O jogo terminou empatado.")

print("=" * 60)
```
