```python
import random
import socket

# ==========================================================
# IDENTIFICACAO DO COMPUTADOR
# ==========================================================

laptop = socket.gethostname()

print("=" * 55)
print(f"   DESAFIO DE PROGRAMACAO - {laptop}")
print("=" * 55)


# ==========================================================
# JOGO 1 - ADIVINHE O NUMERO
# ==========================================================

print("\n🎯 JOGO 1 - ADIVINHE O NUMERO")
print("Acerte o numero para desbloquear o Jogo da Forca!")
print("-" * 55)

# Computador sorteia um numero entre 1 e 20
segredo = random.randint(1, 20)

# Controla se o jogador venceu
venceu_numero = false

print("Adivinhe o numero de 1 a 20.")
print("Voce tem 5 tentativas.")

for tentativa in range(1, 6):

    palpite = int(
        input(f"\nTentativa {tentativa}/5 - seu numero: ")
    )

    if palpite == segredo:
        print("\n🎉 ACERTOU! PARABENS!")
        venceu_numero = True
        break

    elif palpite < segredo:
        print("⬆️ O numero e MAIOR.")

    else:
        print("⬇️ O numero e MENOR.")


# ==========================================================
# VERIFICA SE A FORCA FOI DESBLOQUEADA
# ==========================================================

if venceu_numero:

    print("\n" + "=" * 55)
    print("🔓 JOGO DA FORCA DESBLOQUEADO!")
    print("=" * 55)


    # ======================================================
    # JOGO 2 - FORCA
    # ======================================================

    palavras = [
        "python",
        "jupyter",
        "codigo",
        "aegishub",
        "notebook"
    ]

    # Sorteia uma palavra
    palavra = random.choice(palavras)

    # Letras que o jogador acertou
    acertadas = set()

    # Quantidade de erros
    erros = 0

    # Limite de erros
    max_erros = 6

    print("\n💻 JOGO 2 - FORCA")
    print(f"Voce pode errar {max_erros} vezes.")
    print("Descubra a palavra secreta!")

    while erros < max_erros:

        # Monta a palavra mostrando apenas letras descobertas
        mostra = " ".join(
            [c if c in acertadas else "_" for c in palavra]
        )

        print("\nPalavra:", mostra)
        print(f"Erros: {erros}/{max_erros}")

        # Verifica se descobriu todas as letras
        if all(c in acertadas for c in palavra):
            print("\n🏆 PARABENS!")
            print("Voce adivinhou a palavra:", palavra.upper())
            print("VOCE VENCEU OS DOIS DESAFIOS!")
            break

        # Jogador escolhe uma letra
        letra = input("Diga uma letra: ").strip().lower()

        # Ignora entrada vazia
        if not letra:
            continue

        # Considera apenas a primeira letra digitada
        letra = letra[0]

        if letra in palavra:
            acertadas.add(letra)
            print("✅ Boa! Tem essa letra.")

        else:
            erros += 1
            print("❌ Nao tem essa letra.")

    else:
        print("\n💀 FIM DE JOGO!")
        print("A palavra era:", palavra.upper())


# ==========================================================
# SE NAO ACERTOU O NUMERO
# ==========================================================

else:

    print("\n" + "=" * 55)
    print("❌ ACABARAM AS 5 TENTATIVAS!")
    print(f"O numero correto era: {segredo}")
    print()
    print("🔒 JOGO DA FORCA BLOQUEADO!")
    print("Para jogar a Forca, primeiro voce precisa")
    print("vencer o desafio Adivinhe o Numero.")
    print()
    print("Execute novamente a celula para tentar outra vez!")
    print("=" * 55)
```
